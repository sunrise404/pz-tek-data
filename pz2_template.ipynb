## 6. Преобразование типов и расчёты
Функция `to_float` переводит текст с десятичной запятой в число, а пустую строку в `None`.

```python
def to_float(s):
    # текст с десятичной запятой -> число; пустая строка -> None
    s = s.strip()
    if s == "":
        return None
    return float(s.replace(",", "."))
```

## Шпаргалка

| Конструкция | Пример |
|---|---|
| Разделить строку | `"a;b".split(";")` → `['a', 'b']` |
| Текст в число | `int("1")`, `float("7597,3".replace(",", "."))` |
| Длина, сумма, максимум | `len(x)`, `sum(x)`, `max(x)` |
| Отбор из списка | `[v for v in values if v is not None]` |
| Значение из словаря | `row["object_name"]`, `row.get("col", "нет")` |
| Счётчик | `d[k] = d.get(k, 0) + 1` |
| Уникальные значения | `set(...)` |
| Чтение CSV | `csv.DictReader(f, delimiter=";")` |

| Ошибка | Причина | Что делать |
|---|---|---|
| `UnicodeDecodeError: 'utf-8' codec…` | Файл в кодировке cp1251 | `open(path, encoding="cp1251")` |
| `ValueError: could not convert string to float` | Запятая в числе | `.replace(",", ".")` |
| `7597,37597,3` вместо числа | Строку умножили на число | Сначала преобразовать в `float` |
| `TypeError: … 'NoneType'` | В списке есть пропуск | Отфильтровать `None` |
| `KeyError` | Нет такого столбца | Проверить `reader.fieldnames` или использовать `.get()` |
| `AttributeError: 'float' object has no attribute 'strip'` | Ячейку преобразования запустили второй раз | Заново выполнить чтение файла |
| `FileNotFoundError` | Неверная фамилия или папка | Проверить `SURNAME` и выполнить ячейку 1 |
| Неверный период | Даты ДД.ММ.ГГГГ сравниваются как текст | Проверить формат даты |
