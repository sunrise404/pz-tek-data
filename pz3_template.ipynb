## Шпаргалка

| Действие | Пример |
|---|---|
| Создать массив | `np.array([1, 2, 3])`, `np.arange(24)` |
| Атрибуты | `x.ndim`, `x.shape`, `x.size`, `x.dtype` |
| Сменить тип | `x.astype(float)`, `np.char.replace(s, ",", ".").astype(float)` |
| Сменить форму | `x.reshape(90, 24)`, `x.reshape(-1, 24)` |
| Показатели | `x.sum()`, `x.mean()`, `x.max()`, `x.argmax()` |
| С пропусками | `np.nansum(x)`, `np.nanmean(x)`, `np.nanmax(x)`, `np.isnan(x).sum()` |
| Разности соседних | `np.diff(x)` |
| Столбец таблицы в массив | `df["col"].to_numpy()` |

| Ошибка | Причина | Что делать |
|---|---|---|
| `ufunc 'add' did not contain a loop...` | Массив текстовый (`<U...`) | Перевести в числа `astype(float)` |
| `could not convert string to float: '7597,3'` | Запятая в числе | `np.char.replace(s, ",", ".")` |
| `cannot reshape array of size N into shape` | Произведение размеров не равно N | Проверить `x.size` |
| `operands could not be broadcast together` | Формы массивов не совпадают | Сравнить `shape` обоих массивов |
| `nan` в сумме или среднем | В массиве пропуски | `np.nansum`, `np.nanmean` |
| Пропала дробная часть | `astype(int)` отбрасывает её | `np.round(x)` |
| `NameError: name 'np' is not defined` | Не выполнена ячейка 1 | Выполнить ячейку 1 |
